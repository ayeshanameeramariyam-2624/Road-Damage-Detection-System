```python

from collections import Counter
import os

labels_path = "/content/road_damage_dataset/train/labels"

class_counts = Counter()

for filename in os.listdir(labels_path):
    if filename.endswith(".txt"):
        with open(os.path.join(labels_path, filename), "r") as f:
            for line in f:
                parts = line.strip().split()
                if parts:
                    class_id = int(parts[0])
                    class_counts[class_id] += 1

print("✅ Class annotation counts:")

for class_id, count in sorted(class_counts.items()):
    class_name = data["names"][class_id]
    print(f"Class {class_id} ({class_name}): {count}")
```

```python

from PIL import Image, ImageDraw
import os

images_path = "/content/road_damage_dataset/train/images"
labels_path = "/content/road_damage_dataset/train/labels"

image_files = [
    f for f in os.listdir(images_path)
    if f.lower().endswith((".jpg", ".jpeg", ".png"))
]

for filename in image_files[:3]:
    img = Image.open(os.path.join(images_path, filename)).convert("RGB")
    draw = ImageDraw.Draw(img)

    label_file = os.path.splitext(filename)[0] + ".txt"
    label_path = os.path.join(labels_path, label_file)

    if os.path.exists(label_path):
        for line in open(label_path):
            parts = line.strip().split()

            if len(parts) == 5:
                class_id, x, y, w, h = map(float, parts)

                W, H = img.size

                x1 = int((x - w/2) * W)
                y1 = int((y - h/2) * H)
                x2 = int((x + w/2) * W)
                y2 = int((y + h/2) * H)

                draw.rectangle([x1, y1, x2, y2], width=3)

    display(img.resize((500, 500)))
```

```python

from ultralytics import YOLO
from IPython.display import display
from PIL import Image

model = YOLO("/content/road_damage_training/pothole_model/weights/best.pt")

img_path = "/content/test_images/test/images/" + images[0]

results = model.predict(
    source=img_path,
    conf=0.25,
    save=True
)

print("✅ Detection completed!")

# Show detected image
annotated = results[0].plot()
display(Image.fromarray(annotated[..., ::-1]))
```

```python

r = results[0]

H, W = r.orig_shape

for box in r.boxes:
    cls_id = int(box.cls[0])
    confidence = float(box.conf[0])

    x1, y1, x2, y2 = box.xyxy[0].tolist()

    box_area = (x2 - x1) * (y2 - y1)
    image_area = W * H
    area_ratio = box_area / image_area

    if area_ratio >= 0.15:
        severity = "High"
    elif area_ratio >= 0.05:
        severity = "Medium"
    else:
        severity = "Low"

    print(f"Detected: {model.names[cls_id]}")
    print(f"Confidence: {confidence:.2f}")
    print(f"Severity: {severity}")
```

```python

from ultralytics import YOLO

model = YOLO("/content/road_damage_training/pothole_model/weights/best.pt")

metrics = model.val(
    data="/content/road_damage_dataset/data.yaml",
    split="test",
    imgsz=640
)

print("========== FINAL TEST PERFORMANCE ==========")
print(f"Precision: {metrics.box.mp:.3f}")
print(f"Recall: {metrics.box.mr:.3f}")
print(f"mAP50: {metrics.box.map50:.3f}")
print(f"mAP50-95: {metrics.box.map:.3f}")
```
